# NGSolve tutorial — start here

This is the **in-notebook twin** of the tutorial's landing page: a map of every unit you
can navigate from right here. **Run the cell below** to get a clickable list whose links
open the matching copy for wherever you are — *Google Colab*, *JupyterLite*, or a *local
Jupyter*. The full rendered website (with executed 3D views) lives at
<https://ngsuite.pages.gwdg.de/ngsum2026-tutorial/>.

```python
# Build the tutorial map. Each link points at the *right copy* of the unit for wherever you
# opened this notebook: on Google Colab -> the Colab notebooks; in JupyterLite or a local
# Jupyter -> the sibling .ipynb file (which opens in the same Lab).
import sys
from IPython.display import display, Markdown

_COLAB = "google.colab" in sys.modules
_GH = "schruste/ngsum2026-colab"          # the GitHub mirror Colab opens from


def _link(nb):
    if _COLAB:
        return f"https://colab.research.google.com/github/{_GH}/blob/colab/{nb}.ipynb"
    return f"{nb}.ipynb"                    # JupyterLite & local Jupyter open the sibling


PARTS = [
    ("0 · Setting out — install & run", [
        ("00-setup", "0 · Setting out — install, the five ways to run, a first Beast"),
    ]),
    ("I · Taming the Beast — basics, geometries & first PDEs", [
        ("01-meet-the-beast",        "1 · Meet the Beast — the whole NGSolve loop, walked twice"),
        ("02-geometry",              "2 · Creating Geometry and Meshes"),
        ("03-coefficientfunctions",  "3 · What is a CoefficientFunction?"),
        ("04-fespaces",              "4 · A zoo of finite element spaces"),
        ("05-solving",               "5 · Solving — Dirichlet dofs & static condensation"),
        ("06-linear-solvers",        "6 · The solver toolbox"),
    ]),
    ("II · In the Saddle — features for applications", [
        ("07-saddle-point",            "7 · Mixed problems — the saddle point"),
        ("08-unsteady-doubleglazing",  "8 · Unsteady problems — the double-glazing flow"),
        ("09-nonlinear-allencahn",     "9 · Nonlinear problems — Allen–Cahn & Newton"),
        ("10-dg-hdg",                  "10 · Discontinuous Galerkin & HDG"),
    ]),
    ("III · Taking Flight — advanced techniques", [
        ("11-turing-patterns",         "11 · How the beast got its stripes"),
        ("12-pedestrian-dynamics",     "12 · A crowd heads for coffee"),
        ("13-thermal-plume-hdg",       "13 · A puffing thermal plume"),
        ("14-melting-chocolate",       "14 · Melting the chocolate"),
    ]),
    ("IV · The Grand Expedition — onward to the User Meeting", [
        ("15-outlook-unfitted",        "15 · Outlook — unfitted FEM with ngsxfem (pre-rendered)"),
    ]),
]

# V · Graveyard: slated for removal and NOT built (keeps CI light) — link the sources.
# Own name (not _GH!): reassigning _GH here would clobber the GitHub repo that
# _link() uses, so every Colab link above would wrongly point at GitLab.
_GL = "https://gitlab.gwdg.de/ngsuite/ngsum2026-tutorial/-/blob/main/notebooks/"
GRAVEYARD = [
    ("16-transient-cooling",     "16 · Cooling down in time"),
    ("17-elasticity",            "17 · Making a chocolate bar bend"),
    ("18-buoyant-convection",    "18 · A glimpse of more: buoyant convection"),
    ("19-thermoelasticity",      "19 · A warm chocolate bar bends"),
    ("20-fan-corotating-hdg",    "20 · A spinning fan"),
]

_where = "Google Colab" if _COLAB else "this environment (JupyterLite / local Jupyter)"
_md = [f"### Tutorial map — pick a unit\n",
       f"*Links open the matching copy for **{_where}**.*\n"]
for _head, _items in PARTS:
    _md.append(f"\n**{_head}**\n")
    for _nb, _label in _items:
        _md.append(f"- [{_label}]({_link(_nb)})")
_md.append("\n**V · Graveyard** — *slated for removal; not built — sources on GitLab:*\n")
for _nb, _label in GRAVEYARD:
    _md.append(f"- [{_label}]({_GL}{_nb}.py)")
display(Markdown("\n".join(_md)))
```
