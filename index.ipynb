This is the **in-notebook twin** of the tutorial's landing page: a map of every unit you
can navigate from right here. **Run the cell below** to get a clickable list whose links
open the matching copy for wherever you are — *Google Colab*, *JupyterLite*, or a *local
Jupyter*. The full rendered website (with executed 3D views) lives at
<https://ngsuite.pages.gwdg.de/ngsum2026-tutorial/>.

```python
# Build the tutorial map — the same units, titles and blurbs as the website landing page, each
# with its thumbnail. Links open the right copy for wherever you opened this (Colab -> the Colab
# notebooks; JupyterLite / local -> the sibling .ipynb). Thumbnails embed from data/thumbs when
# present, else fall back to the deployed copy (e.g. on Colab).
import sys, base64, pathlib
from IPython.display import display, Markdown

_COLAB = "google.colab" in sys.modules
_GH = "schruste/ngsum2026-colab"
_SITE = "https://ngsuite.pages.gwdg.de/ngsum2026-tutorial"


def _link(nb):
    if _COLAB:
        return f"https://colab.research.google.com/github/{_GH}/blob/colab/{nb}.ipynb"
    return f"{nb}.ipynb"


def _thumb(nb):
    num = nb.split("-")[0]
    for base in ("data/thumbs", "notebooks/data/thumbs"):     # live notebook / website-build cwd
        p = pathlib.Path(base) / f"thumb-{num}.webp"
        if p.exists():
            return "data:image/webp;base64," + base64.b64encode(p.read_bytes()).decode()
    return f"{_SITE}/static/thumbs/thumb-{num}.webp"           # Colab: the deployed copy


PARTS = [
    ('0 · Setting out — install & run', [
        ('00-setup', '0 · Setting out', '**Start here.** Install NGSolve and explore ways to digest these notebooks.'),
    ]),
    ('I · Taming the Beast — basics, geometries & first PDEs', [
        ('01-meet-the-beast', '1 · Meet the Beast', 'The whole NGSolve loop — geometry & mesh, a variational solve, visualization — on your companion creature.'),
        ('02-geometry', '2 · Creating Geometry and Meshes', 'Forge a sword and a shield, tame the mesh.'),
        ('03-coefficientfunctions', '3 · What is a CoefficientFunction?', 'local/world coordinates, regions, ... and a creative way to implement FEM'),
        ('04-fespaces', '4 · A zoo of finite element spaces', 'Explore some of `NGSolve` finite element spaces.'),
        ('05-solving', '5 · First linear solve', 'Solving linear variational problems and how to treat Dirichlet data.'),
        ('06-linear-solvers', '6 · The solver toolbox 🛠', 'Direct vs iterative solvers, a short glimpse on solvers.'),
    ]),
    ('II · In the Saddle — features for applications', [
        ('07-saddle-point', '7 · Mixed problems — the saddle point 🐎', 'Product FESpaces, block systems, scalar constraints'),
        ('08-unsteady-doubleglazing', '8 · Unsteady problems — the double-glazing flow', 'Time marching with implicit discretizations.'),
        ('09-dg-hdg', '9 · Discontinuous Galerkin & HDG', 'Non-conforming methods & upwind stabilization.'),
        ('10-nonlinear-allencahn', '10 · Nonlinear problems — Allen–Cahn & Newton', "Newton's method on a phase-separation flow."),
    ]),
    ('III · Taking Flight — advanced techniques, and the Grand Expedition', [
        ('11-turing-patterns', '11 · A different pattern: encounter with a relative 🧬', 'Turing patterns on the curved surface of the beast.'),
        ('12-pedestrian-dynamics', '12 · A crowd heads for coffee 🚶☕', 'Pedestrian flow model: the crowd, the coffee and the jam.'),
        ('13-thermal-plume-hdg', '13 · A puffing thermal plume 🔥🌀', 'A buoyant plume: IMEX with Hdiv/HDG'),
        ('14-melting-chocolate', '14 · Melting the chocolate 🍫☕', 'A phase-field Stefan problem: melting chocolate.'),
        ('15-outlook-unfitted', '15 · Deforming domains (ngsxfem) 🫧', 'A two-phase rising bubble on an unfitted mesh (`ngsxfem`).'),
    ]),
]


_where = "Google Colab" if _COLAB else "this environment (JupyterLite / local Jupyter)"
_md = ["### Tutorial map \u2014 pick a unit\n",
       f"*Links open the matching copy for **{_where}**.*\n"]
for _head, _items in PARTS:
    _md.append(f"\n#### {_head}\n")
    for _nb, _title, _desc in _items:
        _md.append(
            f'<img src="{_thumb(_nb)}" alt="" style="float:left;width:68px;height:auto;'
            f'margin:3px 14px 6px 0;border-radius:5px">\n\n'
            f'**[{_title}]({_link(_nb)})** \u2014 {_desc}\n\n'
            f'<div style="clear:both"></div>\n')
display(Markdown("\n".join(_md)))
```
