## 2. The ways the Beast answers — running NGSolve

*"A companion answers in many forms,"* said the master. There are three classic ways to
run an NGSolve computation, plus two that need **no installation at all**.

**(a) The old hall — the Netgen GUI.** Historically you launched scripts *through* Netgen,
which opened its own desktop window with the mesh and solution:

```bash
netgen my_script.py
```

The script draws into the GUI with the same `Draw(...)` calls. This still works and is handy
for quick desktop visualisation, but it needs a graphical display.

```bash
python my_script.py
```

A `Draw(...)` then opens a Netgen viewer window (import `netgen.gui` first to enable it).
Perfect for batch runs, clusters, and reproducible pipelines — no notebook required.

![A terminal running an NGSolve script with the solver's convergence output](https://raw.githubusercontent.com/schruste/ngsum2026-colab/colab/data/run-python-shell.png)

## 3. Proof it lives — conjure the Beast

Let us **summon your future companion** straight away. The **Beast** is NGSolve's logo
**sculpture**: a thick spherical shell pierced by three orthogonal cylindrical bores. We
build it with **Netgen/OCC** constructive geometry (sphere and cylinder primitives
combined with boolean `-`), mesh it, and draw it. If the 3D view appears below, everything
is in place — your companion **is present**, and you are ready for unit 1.

````python
import ngsolve
from netgen.occ import *
from ngsolve import Mesh
from ngsolve.webgui import Draw

print("NGSolve version:", ngsolve.__version__)

shell = Sphere(Pnt(50, 50, 50), 80) - Sphere(Pnt(50, 50, 50), 50)        # the hollow shell
for p, d in [(Pnt(-100, 0, 0), X), (Pnt(100, -100, 100), Y), (Pnt(0, 100, -100), Z)]:
    shell = shell - Cylinder(p, d, r=40, h=300)                          # three bores
beast = shell.Move((-50, -50, -50)).Scale(Pnt(0, 0, 0), 0.05)            # centre + shrink

mesh = Mesh(OCCGeometry(beast).GenerateMesh(maxh=1.0))
mesh.Curve(2)
print(f"the Beast, conjured: {mesh.nv} vertices, {mesh.ne} elements")
Draw(mesh)
````
